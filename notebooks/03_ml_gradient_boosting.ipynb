# 03 - Gradient boosting to predict damage grade
scikit-learn `HistGradientBoostingClassifier`, structure-only vs structure + location. Takes ~1.5 min.

```python
from gfr import ml_model as M
df = pd.read_csv('data/processed/gorkha_clean.csv.gz')
res = M.run(df, cfg)
pd.DataFrame({k: {'micro_f1': v['micro_f1'], 'macro_f1': v['macro_f1'], 'QWK': v['qwk']} for k, v in res.items() if isinstance(v, dict)}).T.round(3)
```
