## Discussion points
1. **Medians (DG>=2 / DS2):** analytical medians are ~1.2x the empirical ones (RC_NE 3+ floors: 0.26-0.29 g vs 0.23 g; RC_E: 0.35-0.41 g vs 0.33 g). **DG3 / DS3:** analytical medians are ~0.63-0.68x the empirical ones (RC_NE 0.50-0.60 g vs 0.88 g; RC_E 0.77-0.91 g vs 1.24 g, and the RC_E empirical DG3 is flagged as extrapolated beyond the surveyed intensity range).
2. **Low-intensity tail:** at 0.15 g the empirical curves predict far more DG>=2 (RC_NE 27 % vs 8 %; RC_E 20 % vs 1 %). The survey mixes infill/soft-storey/poor-quality effects that a bare-frame model cannot produce.
3. **Dispersion:** empirical beta 0.71 (RC_NE 3+) and 0.92 (RC_E 3+) vs analytical beta_total 0.39-0.46. Empirical beta absorbs within-region shaking variability, construction quality and class heterogeneity; analytical beta has only record-to-record (0.25-0.34) plus an assumed 0.30 capacity term.
4. **IM definitions differ:** empirical PGA is inferred and anchored (see notebook 02 sensitivity); analytical PGA comes from synthetic records scaled to PGA.
5. **Model limits:** bare frame (no infill), no pinching / strength degradation, no P-delta, SDOF-based IDA with an equal-energy EPP idealisation, DS thresholds from the pushover (DS2 = effective yield, DS3 = first column hinge at rotation capacity or 4 % drift).
